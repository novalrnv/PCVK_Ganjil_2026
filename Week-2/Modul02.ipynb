##4. Citra yang dibaca dengan skimage.io.imread() ditampilkan berdampingan dengan hasil cv2.cvtColor(img, cv2.COLOR_BGR2RGB) atas citra yang sama.

```python
import cv2 as cv
import matplotlib.pyplot as plt
from skimage import io
import numpy as np

# Baca dengan OpenCV (BGR)
img_bgr = cv.imread('KTM2.jpeg')

# Konversi BGR -> RGB
img_rgb_cv = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

# Baca dengan skimage (RGB)
img_rgb_skimage = io.imread('KTM2.jpeg')

# Tampilkan
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb_skimage)
plt.title("skimage.io.imread() - RGB")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img_bgr)
plt.title("cv2.imread() - BGR\n(Kanal Tertukar)")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(img_rgb_cv)
plt.title("cv.cvtColor() - RGB")
plt.axis("off")

plt.show()

# Ambil satu pixel
y = 100
x = 100

print("Pixel skimage (RGB)       :", img_rgb_skimage[y, x])
print("Pixel cv2.imread (BGR)    :", img_bgr[y, x])
print("Pixel cvtColor (RGB)      :", img_rgb_cv[y, x])
```

##1. Menampilkan citran dengan kanal Merah-Biru dan kanal Hijau-Biru

```python
import cv2
import matplotlib.pyplot as plt
from skimage import io

url = "https://picsum.photos/800/600?random=10"
img = io.imread(url)

# Mengubah BGR menjadi RGB
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Red-Blue
img_RB = imgRGB.copy()
img_RB[:, :, 1] = 0

# Green-Blue
img_GB = imgRGB.copy()
img_GB[:, :, 0] = 0

# Menampilkan berdampingan
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_RB)
plt.title("Kanal Merah-Biru")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_GB)
plt.title("Kanal Hijau-Biru")
plt.axis("off")

plt.tight_layout()
plt.show()
```

##2. Menampilkan citra dengan 3 bentuk pencerminan

```python
import cv2
import matplotlib.pyplot as plt
from skimage import io

url = "https://picsum.photos/800/600?random=10"
img = io.imread(url)

# Konversi BGR ke RGB
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Flip vertikal
flip_vertical = cv2.flip(imgRGB, 0)

# Flip horizontal
flip_horizontal = cv2.flip(imgRGB, 1)

# Flip vertikal dan horizontal
flip_both = cv2.flip(imgRGB, -1)

# Menampilkan ketiganya
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertical)
plt.title("Pencerminan Vertikal")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title("Pencerminan Horizontal")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(flip_both)
plt.title("Pencerminan Vertikal + Horizontal")
plt.axis("off")

plt.tight_layout()
plt.show()
```

#TUGAS PRAKTIKUM - Analisa

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Memilih dan mengupload gambar
uploaded = files.upload()

# Mengambil nama file yang diupload
nama_file = list(uploaded.keys())[0]

# Membaca gambar
img = cv2.imread(nama_file)

# Mengubah gambar menjadi grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Menghitung jumlah seluruh pixel
total_pixel = img_gray.size

# Menghitung pixel dengan intensitas kurang dari 100
pixel_gelap = np.sum(img_gray < 100)

# Menghitung persentase
persentase = (pixel_gelap / total_pixel) * 100

# Menampilkan gambar
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Foto: " + nama_file)
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.tight_layout()
plt.show()

# Menampilkan hasil analisis
print("Nama file              :", nama_file)
print("Ukuran gambar          :", img_gray.shape)
print("Total pixel            :", total_pixel)
print("Pixel dengan intensitas < 100 :", pixel_gelap)
print("Persentase pixel < 100 :", round(persentase, 2), "%")
```

##Studi Kasus 1

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload gambar
uploaded = files.upload()

for nama_file in uploaded.keys():

    img = cv2.imread(nama_file)
    tinggi, lebar = img.shape[:2]

    # Tinggi bar menyesuaikan tinggi gambar
    tinggi_bar = tinggi // 12

    # Membuat bar hitam
    cv2.rectangle(
        img,
        (0, tinggi - tinggi_bar),
        (lebar, tinggi),
        (0, 0, 0),
        -1
    )

    # Menambahkan teks
    cv2.putText(
        img,
        "Dokumentasi PCVK - 2026",
        (10, tinggi - 15),
        cv2.FONT_HERSHEY_SIMPLEX,
        tinggi / 1000,
        (255, 255, 255),
        max(1, tinggi // 300)
    )

    # Tampilkan
    plt.figure(figsize=(8, 6))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title(f"Masukan {lebar} x {tinggi} px | Bar = {tinggi_bar} px")
    plt.axis("off")
    plt.show()
```

##Studi Kasus 2

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload gambar
uploaded = files.upload()

for nama_file in uploaded.keys():

    # Membaca gambar
    img = cv2.imread(nama_file)
    tinggi, lebar = img.shape[:2]

    # Ukuran kanvas persegi
    ukuran = max(tinggi, lebar)

    # Membuat kanvas berwarna putih
    canvas = np.ones((ukuran, ukuran, 3), dtype=np.uint8) * 255

    # Menempatkan gambar di tengah
    x = (ukuran - lebar) // 2
    y = (ukuran - tinggi) // 2

    canvas[y:y+tinggi, x:x+lebar] = img

    cv2.putText(canvas,"NOPAL STORE",(20, ukuran - 20),cv2.FONT_HERSHEY_COMPLEX,
    ukuran / 800,(0, 0, 0),max(2, ukuran // 250))

    # Tampilkan
    plt.figure(figsize=(6, 6))
    plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()
```
